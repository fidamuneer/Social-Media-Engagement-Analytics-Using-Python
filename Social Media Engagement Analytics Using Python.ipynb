Task2 : Data Cleaning

Handling Missing Values

```python
df.isnull().sum()
```

Remove rows only the critical fields are missing. Here the posted_at is the critical field, even though it doesn't have missing values right now.

```python
df["likes"] = df["likes"].fillna(df["likes"].median())
```

```python
df["comments"] = df["comments"].fillna(df["comments"].median())
```

```python
df["shares"] = df["shares"].fillna(df["shares"].median())
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df["gender"] = df["gender"].fillna("Unknown")
```

```python
df["gender"].value_counts()
```

```python
df["sentiment"] = df["sentiment"].fillna("Unknown")
```

```python
df["sentiment"].value_counts()
```

```python
df. isnull().sum()
```

Handling Duplicates

```python
Duplicated_rows = df.duplicated().sum()
Duplicated_rows
```

Data Formatting

```python
df.dtypes
```

```python
numeric_cols = ["age", "likes", "comments", "shares",
                "watch_time_sec", "impression_count",
                "follower_count", "engagement_rate"]

df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors="coerce")
```

```python
df[numeric_cols].dtypes
```

Standardize Categorical Values

```python
categorical_cols = ["country", "post_type", "post_category", "device_type", "sentiment"]
for col in categorical_cols:
    df[col] = df[col].str.strip().str.lower()
```

Checking Unrealistic Values

```python
check_cols = ["likes", "comments", "shares",
              "watch_time_sec", "impression_count", "follower_count"]
(df[check_cols] < 0).sum()
```

```python
df["age"].describe()
```

Extract Hashtag count

```python
df["hashtag_count"] = df["hashtags"].str.count("#")
```

```python
df[["hashtags" , "hashtag_count"]].head(10)
```

```python
df["hashtag_count"] = df["hashtag_count"].fillna(0)
```

Cleaning sentiment label

```python
df["sentiment"] = df["sentiment"].str.strip().str.lower()
```
